```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
dataset=pd.read_csv(r'C:\Users\CC-Lab42\Desktop\Sivaprashanna\AML/iris.csv')
x = dataset.drop('species', axis=1).values
kmeans = KMeans(n_clusters=2, random_state=20)
kmeans.fit(x)
cl = kmeans.labels_
ce = kmeans.cluster_centers_
print("Type of X:", type(x), "Shape of X:", x.shape)
print("Type of cl:", type(cl), "Shape of cl:", cl.shape)
print("Type of ce:", type(ce), "Shape of ce:", ce.shape)

plt.figure(figsize=(10, 6))
plt.scatter(x[:, 0], x[:, 1], c=cl, cmap='viridis', label='Data points')
plt.scatter(ce[:, 0], ce[:, 1], c='black', marker='x', s=200, label='Centroids')
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```
